```python
import pandas as pd
dataset_prep = pd.read_csv(r"/home/srodriguezc/Simulador_de_compra_venta_de_acciones/Data/Merged/Dataset_final.csv")
```

```python
dataset_prep.head()
```

```python
columnas_eliminadas = ['Price', 'High', 'Low', 'Close', 'dd', 'mm', 'yyyy', 'Trade Date', 'Insider Name', 'Title', 'Value', 'title_base', 'title_norm', 'estandar_corr', 'History', 'Stock', 'Activity', 'Share change', '% change to portfolio', 'nombre_ticker', 'ticker', 'fiscalDateEnding', 'quarter_label', 'month', 'reportedEPS', 'estimatedEPS', 'quarter_label_dr', 'Ticker_oi', 'Ticker_dr']

dataset_prep = dataset_prep.drop(columns=columnas_eliminadas, errors='ignore')
```

```python
dataset_prep.head()
```

```python
columnas_orden = ['Date', 'quarter_label_oi', 'Ticker', 'source_file']
# Crear el nuevo orden (primero las definidas, luego el resto en su orden original)
nueva_estructura = columnas_orden + [col for col in dataset_prep.columns if col not in columnas_orden]

# Reordenar sobrescribiendo el mismo DataFrame
dataset_prep = dataset_prep[nueva_estructura]

columnas_cambio_nombre = {'quarter_label_oi': 'quarter - year', 'source_file': 'nombre_holding', 'calificacion' : 'insider_rating', 'surprise' : 'eps_surprise'}

dataset_prep = dataset_prep.rename(columns=columnas_cambio_nombre)

dataset_prep
```

```python
dataset_prep['activity_act'].value_counts()
```

```python
dataset_prep['activity_act'] = dataset_prep['activity_act'].replace({
    'Reduce' : '0',
    'Sell' : '0',
    'Buy' : '1',
    'Add' : '1'
})
dataset_prep['activity_act'] = dataset_prep['activity_act'].astype(int)

dataset_prep['Trade Type'] = dataset_prep['Trade Type'].replace({
    'S' : '0',
    'P' : '1'
})

dataset_prep['Trade Type'] = dataset_prep['Trade Type'].astype(int)

# Reemplazo buy y add por 1, reduce y sell por 0
# cambio el tipo de dato a int
# en la columna Trade Type, reemplazo S por 0 y P por 1, manteniendo consistencia con el cambio en activity_act
```

```python
dataset_prep['diferencia_cierre_vs_apertura'] = dataset_prep['Adj Close'] - dataset_prep['Open']
```

```python
dataset_prep
```

```python
# dejar en el dataset la columna de adj close y eliminar open, close, high, low
#columnas_eliminadas_2 = ['Open','Volume', 'Owned', 'Qty']

columnas_eliminadas_2 = ['Open', 'Adj Close', 'Volume', 'Owned', 'Qty'] # <---- pipeline inicial 
dataset_prep = dataset_prep.drop(columns=columnas_eliminadas_2, errors='ignore')
```

```python
dataset_prep
```

```python
columnas_orden_2 = ['Date', 'quarter - year', 'Ticker', 'diferencia_cierre_vs_apertura', 'nombre_holding', 'activity_act', 'activity_porc']
# Crear el nuevo orden (primero las definidas, luego el resto en su orden original)
nueva_estructura = columnas_orden_2 + [col for col in dataset_prep.columns if col not in columnas_orden_2]

# Reordenar sobrescribiendo el mismo DataFrame
dataset_prep = dataset_prep[nueva_estructura]

columnas_cambio_nombre_2 = {'Trade Type' : 'Operacion_insider', 'ΔOwn' : 'resultado_operacion_insider'} # <----- pipeline inicial

dataset_prep = dataset_prep.rename(columns=columnas_cambio_nombre_2) # <----- pipeline inicial

dataset_prep
```
